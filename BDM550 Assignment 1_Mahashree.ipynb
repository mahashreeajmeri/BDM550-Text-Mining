```python
# 1  (2 marks) Define a conditional frequency distribution over the Names Corpus that allows 
# you to see which initial letters are more frequent for males versus females.

import nltk
from nltk.corpus import names
nltk.corpus.names.fileids()
```

```python
# 2. (3 marks) Consider the attached file as your corpus warlordofmars.txt. Write a program 
# to:
# a) Write a function that finds the 50 most frequently occurring words of a text that are not 
# stopwords.
from nltk.corpus import stopwords
stop_words= set(stopwords.words('english'))

f= open("warlordofmars.txt","r")
text=f.read().lower()
words=text.split()

# remove stopwords
filtered_words= [word for word in words if word not in stop_words]

# count words
fdist= nltk.FreqDist(filtered_words)

# 50 most common words
print(fdist.most_common(50))
```

```python
# b) Write a program to print the 50 most frequent bigrams (pairs of adjacent words) of a 
# text, omitting bigrams that contain stopwords.

bigrams=nltk.bigrams(words)

filtered_bigrams=[(word1, word2) 
                  for word1,word2 in bigrams
                  if word1 not in stop_words and word2 not in stop_words]

# count words
bigram_count= nltk.FreqDist(filtered_bigrams)

# 50 most common bigrams
print(bigram_count.most_common(50))
```

```python
# c) Store the n most likely words in a list words, then randomly choose a word from the list 
# using random.choice(). (You will need to import random first.)

import random
fdist= nltk.FreqDist(filtered_words)
# 50 most frequent words
common_words=[word for word, frequeny in fdist.most_common(50)]

# choose a word randomly
random_word= random.choice(common_words)
print(random_word)
```

```python
# d) Train a model on this corpus and get it to generate random text. You may have to 
# experiment with different start words. How intelligible is the text? Discuss the strengths 
# and weaknesses of this method of generating random text.

bigrams=nltk.bigrams(words)

model={}

for word1, word2 in bigrams:
    if word1 not in model:
        model[word1]=[]

    model[word1].append(word2)

# generate random text
start_word="mars"
generated_words=[start_word]

for i in range(50):
    if generated_words[-1] in model:
        next_word= random.choice(model[generated_words[-1]])
        generated_words.append(next_word)
    else:
        break
print(" ".join(generated_words))
```
